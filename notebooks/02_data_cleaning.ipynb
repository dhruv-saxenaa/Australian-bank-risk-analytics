```python
import pandas as pd
from pathlib import Path
```

```python
raw_folder = Path("../data/raw")
processed_folder = Path("../data/processed")
processed_folder.mkdir(parents=True, exist_ok=True)
```

```python
bank_files = {
    "CBA": "Commonwealth Bank of Australia.csv",
    "NAB": "National Australia Bank.csv",
    "WBC": "Westpac Banking Corporation.csv",
    "ANZ": "Australia and New Zealand Banking Group.csv",
    "MQG": "Macquarie Group Limited.csv",
}
```

```python
banks = {}

for bank, filename in bank_files.items():

    file_path = raw_folder / filename

    df = pd.read_csv(file_path)

    banks[bank] = df

    print(bank, ":", df.shape[0], "rows loaded", sep="")
```

```python
from IPython.display import display

for bank, df in banks.items():

    print("\n" + "=" * 50)
    print(bank)
    print("=" * 50)

    display(df.head())

```

```python
for bank, df in banks.items():

    print("\n", bank)

    print(df.isnull().sum())
```

```python
for bank, df in banks.items():
    print(bank)
    print(df.head())
    print(df.columns)
```

```python
for bank, df in banks.items():

    df["Date"] = pd.to_datetime(df["Date"])

    banks[bank] = df
```

```python
for bank, df in banks.items():

    duplicates = df["Date"].duplicated().sum()

    print(bank, ":", duplicates, "duplicate dates")
```

```python
for bank, df in banks.items():

    df = df.sort_values("Date")

    banks[bank] = df
```

```python

for bank, df in banks.items():
    df.to_csv(
        processed_folder / f"{bank}_clean.csv",
        index=False
    )
print("Cleaned bank files saved successfully.")
```

```python
returns = pd.DataFrame()
for bank, df in banks.items():
    returns[bank] = df['Adj Close'].pct_change()

returns = returns.dropna()
```

```python
import numpy as np
log_returns = np.log(1 + returns)
```

```python
log_returns.to_csv('../data/processed/daily_log_returns.csv')
```
