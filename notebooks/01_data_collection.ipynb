```python
import yfinance as yf 
import pandas as pd
```

```python
tickers = {
    'CBA.AX': 'Commonwealth Bank of Australia',
    'NAB.AX': 'National Australia Bank',
    'WBC.AX': 'Westpac Banking Corporation',
    'ANZ.AX': 'Australia and New Zealand Banking Group',
    'MQG.AX': 'Macquarie Group Limited',
}
```

```python
start_date = '2001-07-30'
end_date = '2026-07-29'
all_data = {}
```

```python
for ticker, name in tickers.items():

    data = yf.download(
        ticker,
        start=start_date,
        end=end_date,
        auto_adjust=False
    )
    
    if isinstance(data.columns, pd.MultiIndex):
        data.columns = data.columns.get_level_values(0)

    all_data[name] = data
```

```python
for bank, df in all_data.items():

    print("\n" + bank)
    print("First date:", df.index.min())
    print("Last date :", df.index.max())
```

```python
for name, df in all_data.items():

    print(name, ":", df.shape[0], "rows downloaded")
```

```python
for name, df in all_data.items():

    df.to_csv(f'../data/raw/{name}.csv')
```
